## A CSV File Is Just Plain Text

*Context: No special CSV module needed just to look at it — commas and newlines are all a CSV file actually is. Python's built-in csv module exists for parsing it into rows automatically, but that's a tool for later, not today's point.*

*Input:*


```python
csv_content = "name,count,department\nStapler,12,Office\nPaper,200,Office\n"

with open("supply.csv", "w") as f:
    f.write(csv_content)

with open("supply.csv") as f:
    print(f.read())
```
