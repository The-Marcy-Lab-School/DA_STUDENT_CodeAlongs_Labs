## Scenario

No template today. You're handed a short, AI-generated script that combines a class, a loop, and functions — it runs without crashing, but it's quietly wrong. Your job: trace it by hand, find at least 3 planted logic errors, fix them, and refactor one repeated block into its own function — all independently.


## The Script, Part 1: The Class

*Context: Same `BoardGame` shape from Days 13-14 — nothing about the class signature itself is new.*

*Input:*


```python
class BoardGame:
    def __init__(self, title, players_available, category):
        self.title = title
        self.players_available = players_available
        self.category = category

    def is_available(self):
        return self.players_available >= 0
```

## The Script, Part 2: The Data

*Context: Four games, four categories — worth noting exactly how each one is capitalized before the next part runs.*

*Input:*


```python
games = [
    BoardGame("Catan", 3, "Strategy"),
    BoardGame("Uno", 0, "Card"),
    BoardGame("Chess", 2, "strategy"),
    BoardGame("Sorry!", 4, "Strategy"),
]
```

## The Script, Part 3: Counting Strategy Games

*Context: This loop is meant to count how many games are in the "Strategy" category.*

*Input:*


```python
strategy_count = 0
for i in range(len(games) - 1):
    if games[i].category == "Strategy":
        strategy_count += 1

print("--- Summary ---")
print("Category: Strategy")
print("Count:", strategy_count)
```

## The Script, Part 4: Counting Available Games

*Context: This loop is meant to count how many games currently have at least one player slot open.*

*Input:*


```python
available_count = 0
for game in games:
    if game.is_available():
        available_count += 1

print("--- Summary ---")
print("Category: Available")
print("Count:", available_count)
```

## Fix 1: The Class

*Context: Bug 1: `is_available` used `>=` instead of `>` — a game with 0 players open counted as available.*

*Input:*


```python
class BoardGame:
    def __init__(self, title, players_available, category):
        self.title = title
        self.players_available = players_available
        self.category = category

    def is_available(self):
        return self.players_available > 0
```

## Fix 1, Continued: Rebuilding the Objects

*Context: Real Python gotcha: Redefining a class does NOT retroactively fix objects already built from the old one — they still point to the old class. The `games` list has to be rebuilt after the class is fixed.*

*Input:*


```python
games = [
    BoardGame("Catan", 3, "Strategy"),
    BoardGame("Uno", 0, "Card"),
    BoardGame("Chess", 2, "strategy"),
    BoardGame("Sorry!", 4, "Strategy"),
]
```

## Fix 2: Counting Strategy Games, Refactored

*Context: Bug 2: `range(len(games) - 1)` silently dropped the last game. Bug 3: `== "Strategy"` missed Chess's lowercase `"strategy"`.*

*Input:*


```python
def print_summary(label, count):
    print("--- Summary ---")
    print("Category:", label)
    print("Count:", count)

strategy_count = 0
for game in games:
    if game.category.lower() == "strategy":
        strategy_count += 1

print_summary("Strategy", strategy_count)
```

## Fix 3: Counting Available Games, Refactored

*Context: Same `print_summary` from the last slide, reused — that's the refactor: one function, called twice, instead of two duplicated print blocks.*

*Input:*


```python
available_count = 0
for game in games:
    if game.is_available():
        available_count += 1

print_summary("Available", available_count)
```
