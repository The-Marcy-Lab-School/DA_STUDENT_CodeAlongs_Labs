## The Cost of Checking a List **Adds Up**

*⚠️ Not auto-validated: Intentionally scans a 2,000,000-element list to demonstrate the real O(n) cost directly, not just describe it - not run by the automated validator to avoid slow/flaky timing checks on every build.*

*Input:*


```python
# NOTE: not auto-validated - Intentionally scans a 2,000,000-element list to demonstrate the real O(n) cost directly, not just describe it - not run by the automated validator to avoid slow/flaky timing checks on every build.
import time

# Run this and see the cost of a full scan in real time
user_id_list = list(range(2_000_000))

start = time.time()
found = 1_999_999 in user_id_list   # nearly a full scan
elapsed = time.time() - start

print(f"Found: {found}")
print(f"Took {elapsed:.4f} sec to check membership")
```
